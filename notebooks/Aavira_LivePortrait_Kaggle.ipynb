```python
import shutil, subprocess, sys
from pathlib import Path
DEVICE = "cuda"  # "cpu" for an explicit CPU test
print("Notebook Python:", sys.version)
if DEVICE == "cuda":
    subprocess.run(["nvidia-smi"], check=True)
ROOT = Path("/kaggle/working" if Path("/kaggle/working").is_dir() else "/content")
assert ROOT.is_dir(), "Use Kaggle or Colab, or run the CLI from your checkout." 
```

Use an existing checkout by setting PROJECT, or clone the development branch after it is pushed. Setup requires Internet for public dependencies and model files.

```python
PROJECT = ROOT / "creator-os"
REF = "codex/liveportrait-runtime-validation"
if not PROJECT.exists():
    subprocess.run(["git", "clone", "--branch", REF, "--single-branch",
                    "https://github.com/mathewshebin21-design/aavira-menon-creator-os.git", str(PROJECT)], check=True)
assert (PROJECT / "creator_os/liveportrait.py").is_file(), "Point PROJECT at the updated checkout."
if not shutil.which("uv"):
    subprocess.run([sys.executable, "-m", "pip", "install", "uv==0.9.8"], check=True)
WORK = ROOT / "liveportrait-research"
CMD = [sys.executable, "-m", "creator_os.liveportrait", "--work", str(WORK), "--device", DEVICE]
subprocess.run(CMD + ["runtime"], cwd=PROJECT, check=True)
```

```python
subprocess.run(CMD + ["setup"], cwd=PROJECT, check=True)
```
