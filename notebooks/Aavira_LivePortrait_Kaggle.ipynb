```python
import os, subprocess
if not os.path.exists('/kaggle/working/LivePortrait'):
    subprocess.run(['git','clone','--depth','1','https://github.com/KwaiVGI/LivePortrait.git','/kaggle/working/LivePortrait'],check=True)
subprocess.run(['pip','install','-r','/kaggle/working/LivePortrait/requirements.txt'],check=True)
subprocess.run(['pip','install','huggingface_hub'],check=True)
```

```python
from huggingface_hub import snapshot_download
snapshot_download(repo_id='KwaiVGI/LivePortrait', local_dir='/kaggle/working/LivePortrait/pretrained_weights')
print('Check the downloaded model directory structure against upstream LivePortrait README before inference.')
```

```python
from pathlib import Path
roots=[Path('/kaggle/input'),Path('/kaggle/working')]
photos=[p for root in roots for p in root.rglob('*') if p.is_file() and p.suffix.lower() in ('.jpg','.jpeg','.png') and 'LivePortrait' not in str(p)]
videos=[p for root in roots for p in root.rglob('*') if p.is_file() and p.suffix.lower() in ('.mp4','.mov') and 'LivePortrait' not in str(p)]
print('Source image candidates:',photos[:15]);print('Driving video candidates:',videos[:15])
assert photos and videos, 'Upload both a portrait and a driving video to Kaggle inputs.'
source_image=photos[0];driving_video=videos[0]
print('Selected:',source_image,driving_video,'— verify these before proceeding')
```

```python
import subprocess
subprocess.run(['python','inference.py','-s',str(source_image),'-d',str(driving_video)],cwd='/kaggle/working/LivePortrait',check=True)
print('Inspect output under /kaggle/working/LivePortrait/animations')
```
